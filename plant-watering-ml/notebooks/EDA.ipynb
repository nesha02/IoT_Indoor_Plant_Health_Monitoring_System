```python
import pandas as pd

df = pd.read_csv("../data/processed/dataset.csv")

df.head()

```

```python
df["soil_pct"].value_counts().head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))
df["soil_pct"].hist(bins=30)
plt.title("Soil Moisture Distribution")
plt.xlabel("Soil %")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="plant_type", y="soil_pct", data=df)
plt.title("Soil Moisture per Plant Type")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df["temperature"].hist(bins=30)
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df["humidity"].hist(bins=30)
plt.title("Humidity Distribution")
plt.xlabel("Humidity")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Feature Correlation")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["temperature"], df["soil_pct"], alpha=0.3)
plt.xlabel("Temperature")
plt.ylabel("Soil %")
plt.title("Soil Moisture vs Temperature")
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("../data/final/labeled_dataset.csv")

df.head()
```

```python
df["water_needed"].value_counts()
```

```python
# Distribution per plant
pd.crosstab(df["plant_type"], df["water_needed"])
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))

sns.countplot(
    x="plant_type",
    hue="water_needed",
    data=df
)

plt.title("Plant-wise Water Needed Distribution")
plt.xlabel("Plant Type")
plt.ylabel("Count")

# Better legend labels
plt.legend(title="Water Needed", labels=["No (0)", "Yes (1)"])

plt.show()
```
